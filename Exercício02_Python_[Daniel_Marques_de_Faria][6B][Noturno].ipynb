## Exercício 01: Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.


```python
import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)
print("Integral:", resultado)
```

## Exercício 02: Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.


```python
import numpy as np
from scipy.integrate import solve_ivp

def equacao(x, y):
    return x - y

pontos = np.linspace(0, 10, 100)
solucao = solve_ivp(equacao, [0, 10], [1], t_eval=pontos)

print(solucao.y[0])
```

## Exercício 03: Encontre as raízes da função f(x) = x⁴ - 3x³ + 2 utilizando SciPy.


```python
from scipy.optimize import brentq

def f(x):
    return x**4 - 3*x**3 + 2

raiz1 = brentq(f, 0.5, 1.5)
raiz2 = brentq(f, 2, 3)

print("Raiz 1:", raiz1)
print("Raiz 2:", raiz2)
```

## Exercício 04: Use a função de otimização de SciPy para maximizar f(x) = -x² + 4x + 1.


```python
from scipy.optimize import minimize_scalar

def f(x):
    return -x**2 + 4*x + 1

resultado = minimize_scalar(lambda x: -f(x))

print("x do máximo:", resultado.x)
print("Valor máximo:", f(resultado.x))
```

## Exercício 05: Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).


```python
import numpy as np
from scipy.fft import fft

t = np.linspace(0, 1, 500)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
transformada = fft(sinal)

print(transformada)
```

## Exercício 07: Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0), (1, 2), (2, 3), (3, 5), (4, 4).

## Exercício 08: Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6.


```python
import numpy as np
from scipy.linalg import solve

A = np.array([[3, 2], [1, 4]])
b = np.array([5, 6])

resultado = solve(A, b)
print("x =", resultado[0])
print("y =", resultado[1])
```

## Exercício 09: Ajuste uma curva quadrática aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8).


```python
import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def quadratica(x, a, b, c):
    return a * x**2 + b * x + c

coeficientes, _ = curve_fit(quadratica, x, y)

a, b, c = coeficientes

print("a =", a)
print("b =", b)
print("c =", c)

print(f"Equação ajustada: y = {a:.4f}x² + {b:.4f}x + {c:.4f}")
```

## Exercício 10: Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [0, 2] para x e [0, 1] para y.


```python
from scipy.integrate import dblquad

def funcao(y, x):
    return x * y + x**2

resultado, erro = dblquad(
    funcao,
    0, 2,
    lambda x: 0,
    lambda x: 1
)

print("Valor da integral dupla:", resultado)
print("Estimativa do erro:", erro)
```

## Exercício 19: Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.



```python
import numpy as np
from scipy.fft import fft, fftfreq

t = np.linspace(0, 1, 100)
sinal = (
    np.sin(2 * np.pi * 5 * t)
    + np.cos(2 * np.pi * 10 * t)
)

fourier = fft(sinal)
frequencias = fftfreq(
    len(sinal),
    t[1] - t[0]
)

print("Transformada de Fourier:")
print(fourier)

print("Frequências:")
print(frequencias)
```

## Exercício 06: Plote dois gráficos de linha em subplots diferentes: y₁ = eˣ e y₂ = log(x).


```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.linspace(-2, 2, 100)
y1 = np.exp(x1)

x2 = np.linspace(0.1, 10, 100)
y2 = np.log(x2)

plt.subplot(1, 2, 1)
plt.plot(x1, y1)
plt.title("y = eˣ")

plt.subplot(1, 2, 2)
plt.plot(x2, y2)
plt.title("y = log(x)")

plt.show()
```

## Exercício 08: Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função y = sin(x).


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.3)
plt.show()
```

## Exercício 10: Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = 0.5*np.sin(2*x)

plt.plot(x, y1, color="blue", linestyle="-")
plt.plot(x, y2, color="red", linestyle="--")
plt.plot(x, y3, color="green", linestyle=":")
plt.show()
```

## Exercício 11: Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.


```python
import numpy as np
import matplotlib.pyplot as plt

categorias = ["A", "B", "C"]
grupo1 = np.array([4, 6, 5])
grupo2 = np.array([3, 2, 4])
grupo3 = np.array([2, 4, 3])

plt.bar(categorias, grupo1, label="Grupo 1")
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")
plt.bar(categorias, grupo3, bottom=grupo1 + grupo2, label="Grupo 3")
plt.legend()
plt.show()
```

## Exercício 18: Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.


```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E", "F"]
valores = [12, -7, 9, -4, 15, -10]
cores = ["green" if valor >= 0 else "red" for valor in valores]

plt.bar(categorias, valores, color=cores)
plt.axhline(0, color="black")
plt.show()
```

## Exercício 19: Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.
