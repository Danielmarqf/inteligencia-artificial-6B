#Exercicio 20 - Matriz Transposta

#Objetivo - Criar a matriz transposta de uma matriz.

#Explicação: Aqui criei uma matriz com três linhas e três colunas e depois criei uma lista vazia para armazenar a matriz transposta.
```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposta = []

for coluna in range(len(matriz[0])):
    nova_linha = []

    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])

    transposta.append(nova_linha)

print("Matriz original:")
print(matriz)

print("Matriz transposta:")
print(transposta)
```

#Exercicio 21 - Busca Binária

#Objetivo - Procurar um número dentro de uma lista ordenada.

#Explicação: Aqui criei uma lista de números já organizada em ordem crescente e um input para que o usuário informe o número que deseja procurar. Depois utilizei as variáveis inicio e fim para definir os limites da busca e um laço while para continuar procurando enquanto ainda existir uma parte da lista para verificar. A variável meio representa a posição central da parte analisada e, dependendo do valor encontrado, o programa diminui ou aumenta o intervalo de busca. Caso o número seja encontrado, o programa informa o resultado e encerra a busca.


```python
lista = [1, 3, 5, 7, 9, 11, 13, 15, 17]

num = int(input("Digite o número que deseja procurar: "))

inicio = 0
fim = len(lista) - 1

while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == num:
        print("Número encontrado!")
        break

    elif num > lista[meio]:
        inicio = meio + 1

    else:
        fim = meio - 1

else:
    print("Número não encontrado.")
```

#Exercicio 22 - Criptografia

#Objetivo - Criptografar uma mensagem utilizando um deslocamento de letras.

#Explicação: Aqui criei uma função chamada criptografar() que recebe uma mensagem e um valor de deslocamento. Utilizei um laço for para percorrer cada letra da mensagem e verifiquei com isalpha() se o caractere é uma letra. Depois utilizei ord() e chr() para transformar as letras em números e realizar o deslocamento dentro do alfabeto. Os caracteres que não são letras são mantidos sem alteração. Por fim, pedi a mensagem e o deslocamento ao usuário e mostrei a mensagem criptografada.


```python
def criptografar(mensagem, deslocamento):
    resultado = ""

    for letra in mensagem:
        if letra.isalpha():
            nova_letra = chr((ord(letra) - ord('a') + deslocamento) % 26 + ord('a'))
            resultado += nova_letra
        else:
            resultado += letra

    return resultado


mensagem = input("Digite uma mensagem: ").lower()
deslocamento = int(input("Digite o deslocamento: "))

mensagem_criptografada = criptografar(mensagem, deslocamento)

print("Mensagem criptografada:", mensagem_criptografada)
```

#Explicação: Aqui criei uma função chamada descriptografar() que recebe a mensagem criptografada e o valor de deslocamento. #Exercicio 24 - Validação de CPF

#Objetivo - Verificar se um CPF é matematicamente válido.

#Explicação: Aqui criei uma função chamada validar_cpf() para verificar se um CPF possui um formato válido e se os seus dígitos verificadores estão corretos. Primeiro removi os pontos e o hífen do CPF informado e depois verifiquei se ele possui exatamente 11 números. Em seguida realizei os cálculos dos dois dígitos verificadores e comparei os resultados com os dígitos informados no CPF. Por fim, utilizei um if/else para mostrar se o CPF informado é válido ou inválido.


```python
def validar_cpf(cpf):
    cpf_limpo = cpf.replace(".", "").replace("-", "")

    if not cpf_limpo.isdigit() or len(cpf_limpo) != 11:
        return False

    if cpf_limpo == cpf_limpo[0] * 11:
        return False

    soma = sum(int(cpf_limpo[i]) * (10 - i) for i in range(9))
    digito_1 = (soma * 10) % 11

    if digito_1 == 10:
        digito_1 = 0

    if digito_1 != int(cpf_limpo[9]):
        return False

    soma = sum(int(cpf_limpo[i]) * (11 - i) for i in range(10))
    digito_2 = (soma * 10) % 11

    if digito_2 == 10:
        digito_2 = 0

    if digito_2 != int(cpf_limpo[10]):
        return False

    return True


cpf_digitado = input("Digite o CPF: ")

if validar_cpf(cpf_digitado):
    print("O CPF informado é válido.")
else:
    print("O CPF informado é inválido.")
```

#Exercicio 25 - Classificação de Números

#Objetivo - Separar números positivos, negativos e zeros.

#Explicação: Aqui criei uma função chamada classificar_numeros() que recebe uma lista de números e cria três listas diferentes para armazenar os valores positivos, negativos e iguais a zero. Depois utilizei um laço for com condições if, elif e else para verificar cada número e adicionar o valor na lista correspondente. Por fim, a função retorna um dicionário com as três categorias e o resultado é mostrado na tela.


```python
def classificar_numeros(lista_numeros):
    positivos = []
    negativos = []
    zeros = []

    for numero in lista_numeros:
        if numero > 0:
            positivos.append(numero)
        elif numero < 0:
            negativos.append(numero)
        else:
            zeros.append(numero)

    return {
        "positivos": positivos,
        "negativos": negativos,
        "zeros": zeros
    }


numeros_misturados = [10, -5, 0, 3.14, -42, 0, 7, -1, 0.0]

resultado = classificar_numeros(numeros_misturados)

print("Resultado da Classificação:")
print(f"Positivos: {resultado['positivos']}")
print(f"Negativos: {resultado['negativos']}")
print(f"Zeros: {resultado['zeros']}")
```

# Exercicio 30 - Jogo da Forca

# Objetivo - Criar um jogo simples da forca permitindo que o usuário tente uma letra ou a palavra completa.

# Explicação: Aqui importei a biblioteca random para escolher aleatoriamente uma palavra de uma lista. Depois criei uma lista de espaços para representar as letras ainda não descobertas e utilizei um laço while para controlar as tentativas do jogador. A cada rodada o usuário pode digitar uma letra ou tentar escrever a palavra inteira. Se digitar uma letra, o programa verifica se ela existe na palavra e mostra as posições corretas, caso contrário uma tentativa é retirada. Se o usuário digitar a palavra completa e acertar, o jogo termina imediatamente com vitória. O jogo também termina quando todas as letras forem descobertas ou quando as tentativas acabarem.


```python
import random

def jogar_forca():
    palavras = ["python", "programacao", "logica", "algoritmo", "computador"]
    palavra_secreta = random.choice(palavras).lower()

    letras_descobertas = ["_" for _ in palavra_secreta]
    tentativas_restantes = 6
    letras_tentadas = []

    print("Jogo da forca")

    while tentativas_restantes > 0 and "_" in letras_descobertas:
        print(f"\nPalavra: {' '.join(letras_descobertas)}")
        print(f"\nTentativas restantes: {tentativas_restantes}")
        print(f"Letras já tentadas: {', '.join(letras_tentadas)}")

        tentativa = input("Digite uma letra ou tente a palavra completa: ").lower().strip()

        if tentativa == palavra_secreta:
            print(f"\nParabéns! Você acertou a palavra: {palavra_secreta}")
            return

        if len(tentativa) > 1:
            tentativas_restantes -= 1
            print("A palavra digitada está errada.")
            continue

        if not tentativa.isalpha():
            print("Digite apenas letras.")
            continue

        if tentativa in letras_tentadas:
            print("Você já tentou essa letra.")
            continue

        letras_tentadas.append(tentativa)

        if tentativa in palavra_secreta:
            print(f"A letra '{tentativa}' está na palavra.")

            for i in range(len(palavra_secreta)):
                if palavra_secreta[i] == tentativa:
                    letras_descobertas[i] = tentativa

        else:
            tentativas_restantes -= 1
            print(f"A letra '{tentativa}' não está na palavra.")

    if "_" not in letras_descobertas:
        print(f"\nParabéns! Você venceu! A palavra era: {palavra_secreta}")
    else:
        print(f"\nGame Over! Você perdeu. A palavra era: {palavra_secreta}")


jogar_forca()
```

#Exercicio 31 - Bubble Sort

#Objetivo - Ordenar uma lista de números utilizando o método Bubble Sort.

#Explicação: Aqui criei uma função chamada bubble_sort() para organizar os números de uma lista em ordem crescente. O programa percorre a lista comparando cada elemento com o próximo e realiza uma troca quando o primeiro é maior que o segundo. Esse processo é repetido até que todos os números estejam na ordem correta. Também utilizei a variável trocou para verificar se alguma troca aconteceu e interromper o processo caso a lista já esteja ordenada.


```python
def bubble_sort(lista):
    n = len(lista)

    for i in range(n):
        trocou = False

        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                trocou = True

        if not trocou:
            break

    return lista


num = [64, 34, 25, 12, 22, 11, 90]

print(f"Lista original: {num}")

num_ordenados = bubble_sort(num)

print(f"Lista ordenada: {num_ordenados}")
```

#Exercicio 32 - Jogo da Velha

#Objetivo - Criar um jogo da velha para dois jogadores.

#Explicação: Aqui criei um jogo da velha utilizando uma lista com nove posições para representar o tabuleiro. Também criei uma função para mostrar o tabuleiro e outra para verificar as combinações que podem resultar em vitória. Durante o jogo, os jogadores escolhem uma posição de 1 a 9 e o programa verifica se a posição está disponível, registra a jogada e alterna entre os jogadores X e O. O jogo termina quando um jogador consegue uma combinação vencedora ou quando todas as posições são preenchidas.


```python
def imprimir_tabuleiro(tabuleiro):
    print()
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")
    print()


def verificar_vitoria(tabuleiro, jogador):
    vitorias = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]

    for condicao in vitorias:
        if tabuleiro[condicao[0]] == tabuleiro[condicao[1]] == tabuleiro[condicao[2]] == jogador:
            return True

    return False


def iniciar_jogo():
    tabuleiro = [" " for _ in range(9)]
    jogador_atual = "X"
    jogadas = 0

    print("Bem-vindo ao Jogo da Velha!")
    print("Use os números de 1 a 9 para escolher sua posição.")

    while True:
        imprimir_tabuleiro(tabuleiro)
        print(f"Vez do jogador {jogador_atual}.")

        try:
            escolha = int(input("Escolha uma posição (1-9): ")) - 1

            if escolha < 0 or escolha > 8:
                print("A posição deve ser um número entre 1 e 9.")
                continue

            if tabuleiro[escolha] != " ":
                print("Essa posição já está ocupada.")
                continue

            tabuleiro[escolha] = jogador_atual
            jogadas += 1

            if verificar_vitoria(tabuleiro, jogador_atual):
                imprimir_tabuleiro(tabuleiro)
                print(f"Parabéns! O jogador {jogador_atual} venceu!")
                break

            if jogadas == 9:
                imprimir_tabuleiro(tabuleiro)
                print("Deu velha.")
                break

            jogador_atual = "O" if jogador_atual == "X" else "X"

        except ValueError:
            print("Digite apenas um número inteiro.")


iniciar_jogo()
```

# Objetivo - Mostrar a sequência de Fibonacci até um número informado pelo usuário.

# Explicação: Aqui criei uma função chamada fibonacci_ate_n() que recebe um número limite informado pelo usuário. Dentro da função criei uma lista vazia chamada sequencia e utilizei as variáveis a e b começando com os valores 0 e 1. Depois utilizei um laço while que continua funcionando enquanto o valor de a for menor ou igual ao número limite. A cada repetição o valor de a é adicionado à lista utilizando append() e depois as variáveis são atualizadas para gerar o próximo número da sequência de Fibonacci. Por fim, a função retorna a sequência e o resultado é mostrado na tela.



```python
def fibonacci_ate_n(n):
    sequencia = []
    a, b = 0, 1

    while a <= n:
        sequencia.append(a)
        a, b = b, a + b
    return sequencia

num_limite = int(input("Digite o número limite para a sequência de Fibonacci: "))

resultado = fibonacci_ate_n(num_limite)

print(f"Fibonacci até {num_limite}:")
print(f"Sequência é até: {resultado}")
```

# Objetivo - Encontrar todos os números primos até um valor informado pelo usuário.

# Explicação: Aqui criei uma função chamada primos_ate_n() que recebe um número limite informado pelo usuário e verifica todos os números a partir de 2 até esse valor. Para cada número criei uma variável chamada n_primo começando como True e utilizei outro laço for para verificar se existe algum divisor. Caso o número seja divisível por algum valor, n_primo passa para False e o laço é interrompido utilizando break. Se o número continuar sendo primo, ele é adicionado à lista utilizando append(). Por fim, a função retorna a lista com todos os números primos encontrados e o resultado é mostrado na tela.



```python
def primos_ate_n(n):
    primos = []

    for num in range(2, n + 1):
        n_primo = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                n_primo = False
                break
        if n_primo:
            primos.append(num)
    return primos

limite = int(input("Digite um número limite: "))

print(f"Números primos até {limite}: {primos_ate_n(limite)}")
```

# Exercicio 43 - Conversão de Comprimento

# Objetivo - Converter valores entre diferentes unidades de comprimento informadas pelo usuário.

# Explicação: Aqui criei uma função chamada converter() que recebe um valor, a unidade de origem e a unidade de destino. Depois criei um dicionário com os fatores de conversão de milímetros, centímetros, metros e quilômetros para metros. Primeiro verifico se as unidades informadas existem no dicionário. Em seguida converto o valor de origem para metros e depois de metros para a unidade de destino. Por fim, criei inputs para que o usuário informe o valor e as unidades desejadas e mostrei o resultado da conversão na tela.


```python
def converter(valor, unidade_origem, unidade_destino):
    fatores_para_metro = {
        "milimetros": 0.001,
        "centimetros": 0.01,
        "metros": 1.0,
        "quilometros": 1000.0
    }

    if unidade_origem not in fatores_para_metro or unidade_destino not in fatores_para_metro:
        return "Unidade não suportada."
    valor_em_metros = valor * fatores_para_metro[unidade_origem]
    valor_convertido = valor_em_metros / fatores_para_metro[unidade_destino]
    return valor_convertido

valor = float(input("Digite o valor que deseja converter: "))

unidade_origem = input(
    "Digite a unidade de origem (milimetros, centimetros, metros ou quilometros): "
).lower()
unidade_destino = input(
    "Digite a unidade de destino (milimetros, centimetros, metros ou quilometros): "
).lower()

resultado = converter(valor, unidade_origem, unidade_destino)

print(f"{valor} {unidade_origem} equivalem a {resultado} {unidade_destino}.")
```

```python
def fibonacci_termos(quantidade_termos):
    if quantidade_termos <= 0:
        return []

    elif quantidade_termos == 1:
        return [0]

    sequencia = [0, 1]

    for i in range(quantidade_termos - 2):
        proximo = sequencia[-1] + sequencia[-2]
        sequencia.append(proximo)
    return sequencia

termos = int(input("Digite quantos termos de Fibonacci deseja mostrar: "))

print(f"Os primeiros {termos} termos de Fibonacci são: {fibonacci_termos(termos)}")
```

# Exercicio 45 - Separação de Pares e Ímpares

# Objetivo - Receber 5 números do usuário e separar os valores pares e ímpares.

# Explicação: Aqui criei uma função chamada classificar_pares_impares() que recebe uma lista de números e separa os valores em duas listas diferentes, uma para os números pares e outra para os números ímpares. Depois criei uma lista vazia chamada numeros e utilizei um laço for para pedir 5 números ao usuário, adicionando cada valor na lista utilizando append(). Em seguida utilizei o operador % para verificar o resto da divisão por 2. Se o resto for igual a zero, o número é considerado par, caso contrário é considerado ímpar. Por fim, mostrei a lista completa e os números separados em pares e ímpares.


```python
def classificar_pares_impares(lista):
    pares = []
    impares = []

    for numero in lista:
        if numero % 2 == 0:
            pares.append(numero)
        else:
            impares.append(numero)

    return pares, impares

numeros = []

for i in range(5):
    numero = int(input(f"Digite o {i + 1}º número: "))
    numeros.append(numero)

pares, impares = classificar_pares_impares(numeros)

print("Lista completa:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)
```

#Exercicio 46 - Média e Desvio Padrão

#Objetivo - Calcular a média e o desvio padrão de uma lista de números.

#Explicação: Aqui importei a biblioteca statistics para utilizar as funções de cálculo estatístico. Criei uma função que recebe uma lista e primeiro verifica se ela está vazia.
```python
import statistics

def calcular_estatisticas(lista):
    if not lista:
        return 0, 0

    media = statistics.mean(lista)
    desvio_padrao = statistics.stdev(lista)

    return media, desvio_padrao


dados = [10, 12, 23, 23, 16, 23, 21, 16]

media, std = calcular_estatisticas(dados)

print("Dados:", dados)
print(f"Média: {media:.2f}")
print(f"Desvio Padrão: {std:.2f}")
```

#Exercicio 48 - Insertion Sort

#Objetivo - Ordenar uma lista de números utilizando o método Insertion Sort.

#Explicação: Aqui criei uma função chamada insertion_sort() para organizar uma lista em ordem crescente. O algoritmo começa pelo segundo elemento da lista e considera que o primeiro já está organizado. Depois compara o elemento atual com os valores anteriores e desloca para a direita aqueles que são maiores, até encontrar a posição correta. Por fim, o elemento é colocado nessa posição e o processo continua até que toda a lista esteja ordenada.


```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1

        while j >= 0 and chave < lista[j]:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista


lista_desordenada = [12, 11, 13, 5, 6]

print("Antes da ordenação:", lista_desordenada.copy())

lista_ordenada = insertion_sort(lista_desordenada)

print("Após a inserção do Insertion Sort:", lista_ordenada)
```

#Exercicio 49 - Soma dos Dígitos

#Objetivo - Somar todos os dígitos de um número.

#Explicação: Aqui criei uma função chamada soma_digitos() que recebe um número e utiliza abs() para trabalhar também com números negativos. Depois converti o número para uma string para conseguir percorrer cada dígito separadamente. Cada caractere é convertido novamente para inteiro e somado através da função sum(). No final a função retorna o resultado da soma dos dígitos.


```python
def soma_digitos(numero):
    numero_absoluto = abs(numero)

    soma = sum(int(digito) for digito in str(numero_absoluto))

    return soma


numero_teste = 4591

resultado = soma_digitos(numero_teste)

print("Soma dos Dígitos:")
print(f"A soma dos dígitos de {numero_teste} é: {resultado}")

numero_neg = -123

print(f"A soma dos dígitos de {numero_neg} é: {soma_digitos(numero_neg)}")
```

# Exercicio 50 - Frequência das Letras

# Objetivo - Contar quantas vezes cada letra aparece em um texto digitado pelo usuário.

# Explicação: Aqui importei Counter da biblioteca collections para facilitar a contagem das letras. Depois criei uma função chamada calcular_frequencia_letras() que recebe um texto e utiliza lower() para transformar todas as letras em minúsculas. Em seguida utilizei isalpha() para manter somente os caracteres que são letras, removendo espaços, números e pontuação. Depois utilizei Counter() para contar quantas vezes cada letra aparece e sorted() para organizar o resultado em ordem alfabética. Por fim, criei um input para que o usuário digite o texto que deseja analisar e utilizei um laço for para mostrar cada letra e a quantidade de vezes que ela apareceu.


```python
from collections import Counter

def calcular_frequencia_letras(texto):
    texto_minusculo = texto.lower()
    apenas_letras = [char for char in texto_minusculo if char.isalpha()]
    frequencia = Counter(apenas_letras)
    return dict(sorted(frequencia.items()))

texto_usuario = input("Digite um texto: ")

frequencia = calcular_frequencia_letras(texto_usuario)

print(f"Texto analisado: '{texto_usuario}'")
print("Frequência das letras:")

for letra, quantidade in frequencia.items():
    print(f"Letra '{letra}': {quantidade} vez(es)")
```
